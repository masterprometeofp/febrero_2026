```python
frutas = ["Manzana", "Pera", "Plátano"]

frutas[1] = "Melocotón"
print(frutas)

frutas.append("Kiwi")
print(frutas)

frutas.insert(1, "Sandia")
print(frutas)

frutas.remove("Kiwi")
print(frutas)

frutas.pop()
print(frutas)

frutas.sort()
print(frutas)

numeros = [2, 45, 3, 45, 6, 23, 9, 98, 21, 43]
numeros.sort()
print(numeros)

numeros.reverse()
print(numeros)

nuevos_numeros = [2, 4, 5, 3, 6, 9, 7, 43, 21, 1, 0, 34, 22]
nuevos_numeros.sort(reverse=True)
print(nuevos_numeros)
```

```python
animales_casa = ["Perro", "Gato", "Pajaro"]
animales_salvajes = ["Leon", "Tigre", "Elefante"]

animales_casa.extend(animales_salvajes)
print(animales_casa)

ultimo = animales_casa.pop()
print(ultimo)
print(animales_casa)

print(animales_casa.index("Gato"))

numeros = [1, 2, 3, 4, 5, 6, 2, 3, 4, 2]
print(numeros.count(2))

lista_original = [1, 2, 3]
lista_copia = lista_original.copy()
lista_copia.append(4)
print(lista_copia)
print(lista_original)
```
