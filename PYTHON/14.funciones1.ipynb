```python
notas_ana = [8, 9, 7]
media_ana = sum(notas_ana) / len(notas_ana)
print(f"Media: {media_ana}")

notas_luis = [6, 5, 7]
media_luis = sum(notas_luis) / len(notas_luis)
print(f"Media: {media_luis}")
```

```python
def saludar():
    print("Hola")
    print("Empezamos con las funciones!")


saludar()
saludar()
```

```python
def saludar(nombre):  # nombre es el parámetro
    print(f"Hola, {nombre}")


saludar("Luis")  # Luis es el argumento para el parametro nombre
saludar("Ana")
```

```python
def presentar(nombre, edad, ciudad):
    print(f"En la ciudad de {ciudad} vive {nombre} con {edad} años de edad")


presentar("Ana", 34, "Cádiz")
```

```python
def mostrar_notas(alumno, notas):
    print(f"Notas de {alumno}")
    for nota in notas:
        print(f"     - {nota}")


mostrar_notas("Ana", [8, 9, 7])
mostrar_notas("Pedro", [7, 6, 2])
```

```python
def calcular_edad(nombre, anioNacimiento, anioActual):
    """
    Esta función sirve para calcular la edad de alguien basada en su año de nacimiento y el año actual

    Args:
        nombre (string): El nombre del usuario al que se le va a calcular la edad
        anioNacimiento (number): El año de nacimiento del usuario
        anioActual (number): El año actual
    """
    edad = anioActual - anioNacimiento
    print(f"{nombre} tiene {edad} años en {anioActual}")


calcular_edad("Byron", 2020, 2026)
```
