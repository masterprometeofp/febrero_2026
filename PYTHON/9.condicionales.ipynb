## Condicionales

```python
temperatura = 26

if temperatura > 30:
    print("Ponte una chaqueta")
```

```python
if temperatura > 30:
    print("Ponte ropa de verano")
else:
    print("Ponte una chaqueta")
```

```python
nota = 8

if nota >= 9:
    print("Sobresaliente")
elif nota >= 7:
    print("Notable")
elif nota >= 5:
    print("Aprobado")
else:
    print("Suspenso")
```

```python
dia = "sábado"

if dia == "sábado" and dia == "domingo":
    print("Es fin de semana")
else:
    print("Es laborable")
```

```python
pais = "España"
edad = 19

if edad >= 18 and (pais == "España" or pais == "Portugal"):
    print("Puedes conducir")

duda = True

if duda:
    print("Resuelvo la duda")

lluvia = False

if not lluvia:
    print("No hace falta paraguas")
```

```python
edad = 20
entrada = True

if edad >= 18:
    if entrada:
        print("Puedes entrar con tu entrada")
    else:
        print("No puedes entrar sin entrada")
else:
    print("No puedes entrar por edad")
```

```python
# Falsy
if 0:
    print("No imprime")
if "":
    print("No imprime")
if []:
    print("No imprime")
if None:
    print("No imprime")

# Truthy
if "hola":
    print("Imprime")
if [1, 2]:
    print("Imprime")
```
