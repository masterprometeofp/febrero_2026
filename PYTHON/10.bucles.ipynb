## Bucles

```python
frutas = ["manzana", "pera", "sandia"]

for fruta in frutas:
    print(fruta)

for letra in "Programación":
    print(letra)

for valor in (10.24567, 12.234534):
    print(valor)
```

```python
for num in range(5):
    print(num)
```

```python
for num in range(2, 8):
    print(num)
```

```python
for num in range(0, 10, 2):
    print(num)
```

```python
for num in range(10, 0, -2):
    print(num)
```

```python
print(list(range(5)))
```

```python
consola = {"nombre": "PlayStation 2", "year": 2003, "marca": "Sony", "color": "Negro"}

for clave, valor in consola.items():
    print(f"La clave {clave} tiene el valor: {valor}")
```

```python
animales = ["Perro", "Gato", "Elefante", "Loro"]

for indice, animal in enumerate(animales, start=1):
    print(f"{indice} - {animal}")
```

```python
nombres = ["Álvaro", "Valeria", "Rubén", "Paco"]
edades = [22, 56, 32, 67]

for nombre, edad in zip(nombres, edades):
    print(f"{nombre} tiene {edad} años.")
```

```python
contador = 0

while contador < 5:
    print(contador)
    contador += 1
```
