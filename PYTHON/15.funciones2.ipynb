```python
# Funciones que llaman a otras funciones
def calcular_media(notas):
    media = sum(notas) / len(notas)
    return media


def obtener_calificacion(media):
    if media >= 9:
        return "Sobresaliente"
    elif media >= 7:
        return "Notable"
    elif media >= 5:
        return "Aprobado"
    else:
        return "Suspenso"


def generar_informe(nombre, notas):
    media = calcular_media(notas)
    calificacion = obtener_calificacion(media)
    return f"{nombre}: {media} de media = {calificacion}"


print(generar_informe("Paco", [4, 6, 8]))
```

```python
# args -> * -> Numero indeterminado de argumentos
def sumar_todo(*numeros):
    print(sum(numeros))


sumar_todo(1, 2, 3, 4, 5)


def presentacion(nombre, *aficiones):
    print(f"{nombre} tiene {len(aficiones)} aficiones:")
    for aficion in aficiones:
        print(f"- {aficion}")


presentacion("Manuel")

presentacion(
    "Antonio",
    "videojuegos",
    "tocar la guitarra",
    "comer",
    "dormir",
    "ver películas",
    "escuchar música",
)
```

```python
# kwargs -> ** -> numero indeterminado de claves y valores -> Los agrupa en un diccionario
def crear_perfil(**datos):
    for clave, valor in datos.items():
        print(f"{clave} : {valor}")


crear_perfil(nombre="Ana", edad=30, ciudad="Madrid")
```

```python
# funciones lambda
def doble(num):
    return num * 2


doble_lambda = lambda num: num * 2


print(doble(4))
print(doble_lambda(4))

calificar = lambda nota: "Aprobado" if nota >= 5 else "Suspenso"

sumar = lambda a, b: a + b
```
