## Funciones built-in 2

```python
# Funciones numéricas

print(abs(-7))
print(abs(2.5))

print(round(3.14159))
print(round(3.14159, 2))

print(pow(2, 10))
```

```python
# Minimo, maximo y sumar
notas = [3, 5, 8]

print(min(notas))
print(max(notas))
print(sum(notas))

print(max(3, 6, 1, 7, 9, 1, 23, 2))
```

```python
palabras = ["Python", "JavaScript", "Go", "Java"]

print(max(palabras, key=len))
print(min(palabras, key=len))

print(sorted(palabras, key=len))
```

```python
print(all([True, True, False]))
print(any([True, True, False]))

print(all([True, True, True]))
print(any([False, False, False]))

print("Ejemplo:")

notas = [7.8, 5.6, 4.6, 2, 8, 9.2]

print(all(nota >= 5 for nota in notas))
print(any(nota >= 5 for nota in notas))
```

```python
print(type(42))
print(type("Hola"))

print(isinstance(42, int))
print(isinstance("Hola", str))
```

```python
# Funciones de ordenamiento
numeros = [3, 5, 7, 6, 23, 1, 43, 8, 42, 9]

print(sorted(numeros))
print(numeros)

print(sorted(numeros, reverse=True))

palabras = ["Python", "JavaScript", "Go", "Java", "Rust"]
print(sorted(palabras, key=len))

numeros_ordenados = sorted(numeros)
print(numeros_ordenados)
```

```python
numeros = [3, 4, 1, 6, 2, 8, 0]

print(numeros)

numeros_ordenados = numeros.copy()
numeros_ordenados.sort()
print(numeros_ordenados)

print(numeros)
```
