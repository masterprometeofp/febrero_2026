```python
# Métodos utiles
frutas = {"manzana", "pera", "sandía"}
print(frutas)
frutas.add("manzana")
frutas.add("melón")
print(frutas)

frutas.update(["kiwi", "tomate", "uvas"])
print(frutas)

# Método tradicional
# frutas.remove("meló")
print(frutas)

# Método seguro
frutas.discard("melón")
frutas.discard("melón")
print(frutas)

print(frutas.pop())
```

```python
# Métodos de conjuntos
terror = {"The Evil Dead", "Barbarian", "Resident Evil", "Smile 2", "Sunshine"}
scifi = {"Sunshine", "Proyecto Salvación", "2001: Odisea en el espacio", "Barbarian"}

# Unión -> | -> Sacar ambos conjuntos sin repeticiones
print(terror | scifi)
print(terror.union(scifi))

# Intersection -> & -> Sacar los elementos que aparecen en ambos conjuntos
print(terror & scifi)
print(terror.intersection(scifi))

# Difference -> - -> Sacar los elementos del a que no están en el b
print(terror - scifi)
print(terror.difference(scifi))

# Symetric Difference -> ^ -> Saca los elementos que estan en uno u otro conjunto pero no en ambos
print(terror ^ scifi)
print(terror.symmetric_difference(scifi))
```

```python
# Métodos de comparación entre conjuntos
small = {1, 2}
big = {1, 2, 3, 4}

# issubset -> ¿El a esta dentro del b?
print(small.issubset(big))
print(small <= big)

# issuperset -> ¿El a engloba al b?
""" print(small.issuperset(big))
print(small >= big) """

print(big.issuperset(small))
print(big >= small)

# isdisjoint -> Preguntamos si los dos conjuntos son unicos y no comparten nada
print(small.isdisjoint(big))
print(small.isdisjoint({6, 7, 8}))
```
