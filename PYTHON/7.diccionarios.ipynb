```python
persona = {"nombre": "Paco", "ciudad": "Madrid", "edad": 50, "dinero": 500}
persona2 = dict(nombre="Paco", ciudad="Madrid", edad=50)
print(persona2)

persona3 = dict()
```

```python
# Acceso a los valores de un diccionario

# Método tradicional:
# print(persona["dinero"])
# Método seguro:
print(persona.get("dinero", "No tiene dinero"))
# De forma adicional al get se le puede añadir una respuesta alternativa si no encuentra la propiedad
print(persona.get("ciudad"))
```

```python
print(persona)

persona["dinero"] = 600
print(persona)

persona["trabajo"] = "Panadero"
print(persona)

del persona["dinero"]
print(persona)

print(persona.pop("edad"))
print(persona.pop("edad", "La propiedad edad no está"))

# persona.clear()
```
