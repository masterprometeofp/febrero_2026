```python
class Perro:
    def __init__(self, nombre, raza, edad):
        self.nombre = nombre
        self.raza = raza
        self.edad = edad

    def ladrar(self):
        print(f"{self.nombre} dice Guau!")

    def edad_humana(self):
        return self.edad * 7


byron = Perro("Byron", "Border-Collie-Galgo", 6)
print(byron.nombre)
byron.ladrar()
print(byron.edad_humana())
```

```python
class Animal:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def comer(self):
        print(f"{self.nombre} está comiendo")

    def presentarse(self):
        print(f"Me llamo {self.nombre} y tengo {self.edad} años")


class Perro(Animal):
    def ladrar(self):
        print(f"{self.nombre} dice Guau!")


class Gato(Animal):
    def maullar(self):
        print(f"{self.nombre} dice Miau!")


ciervo = Animal("Jose Manuel", 4)
ciervo.comer()
ciervo.presentarse()

byron = Perro("Byron", 6)
byron.comer()
byron.presentarse()
byron.ladrar()
```

```python
print(isinstance(byron, Animal))
print(isinstance(ciervo, Animal))
print(issubclass(Perro, Animal))
```
